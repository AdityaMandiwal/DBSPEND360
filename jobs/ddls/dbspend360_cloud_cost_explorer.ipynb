# DDL: `dbspend360_cloud_cost_explorer`

Creates the unified cloud cost table populated by the AWS / Azure / GCP cost explorer ETLs.

**Widgets**
- `catalog` - target Unity Catalog name
- `schema`  - target schema name within `catalog`

```python
dbutils.widgets.text("catalog", "", "Catalog")
dbutils.widgets.text("schema", "", "Schema")
```

```python
catalog = dbutils.widgets.get("catalog").strip()
schema = dbutils.widgets.get("schema").strip()

if not catalog or not schema:
    raise ValueError("Both `catalog` and `schema` widgets must be set.")

spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")
```

```python
%sql
CREATE TABLE IF NOT EXISTS ${catalog}.${schema}.dbspend360_cloud_cost_explorer (
  cluster_id          STRING,
  cloud_cost          DOUBLE,
  compute_cost        DOUBLE,
  storage_cost        DOUBLE,
  network_cost        DOUBLE,
  other_cost          DOUBLE,
  currency            STRING,
  created_at          TIMESTAMP,
  updated_at          TIMESTAMP,
  cost_incurred_date  DATE
)
CLUSTER BY AUTO
```

```python
dbutils.notebook.exit(f"{catalog}.{schema}.dbspend360_cloud_cost_explorer")
```
