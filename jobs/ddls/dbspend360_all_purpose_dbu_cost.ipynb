# DDL: `dbspend360_all_purpose_dbu_cost`

Creates the per-cluster / per-user Databricks DBU cost table populated by `dbspend360_all_purpose_dbu_cost_app`.

Sibling of `dbspend360_dbu_cost`, but keyed on `(cluster_id, user_id, usage_date)` for all-purpose clusters
(`cluster_source IN ('UI','API')`). `user_id` is sourced from `system.compute.clusters.owned_by`; rows where
the owner cannot be resolved are bucketed as `__unknown__` rather than dropped.

**Widgets**
- `catalog` - target Unity Catalog name
- `schema`  - target schema name within `catalog`

```python
dbutils.widgets.text("catalog", "", "Catalog")
dbutils.widgets.text("schema", "", "Schema")
```

```python
catalog = dbutils.widgets.get("catalog").strip()
schema = dbutils.widgets.get("schema").strip()

if not catalog or not schema:
    raise ValueError("Both `catalog` and `schema` widgets must be set.")

spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")
```

```python
%sql
CREATE TABLE IF NOT EXISTS ${catalog}.${schema}.dbspend360_all_purpose_dbu_cost (
  cluster_id          STRING,
  user_id             STRING,
  usage_date          DATE,
  databricks_cost     DOUBLE,
  currency            STRING,
  sku_name            STRING,
  workspace_id        STRING,
  data_security_mode  STRING,
  created_at          TIMESTAMP,
  updated_at          TIMESTAMP
)
CLUSTER BY AUTO
```
