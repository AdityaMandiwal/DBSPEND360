# DDL: `dbspend360_error_log`

Creates the error log table that captures rows rejected during ETL joins / quality checks.

**Widgets**
- `catalog` - target Unity Catalog name
- `schema`  - target schema name within `catalog`

```python
dbutils.widgets.text("catalog", "", "Catalog")
dbutils.widgets.text("schema", "", "Schema")
```

```python
catalog = dbutils.widgets.get("catalog").strip()
schema = dbutils.widgets.get("schema").strip()

if not catalog or not schema:
    raise ValueError("Both `catalog` and `schema` widgets must be set.")

spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")
```

```python
%sql
CREATE TABLE IF NOT EXISTS ${catalog}.${schema}.dbspend360_error_log (
  source_system STRING,   -- 'AWS_COST' or 'AZURE_COST' or 'GCP_COST' or 'DBR_DBU'
  error_type    STRING,   -- e.g., 'NO_MATCH_DBR_DBU', 'NO_MATCH_AWS_COST', 'NO_MATCH_AZURE_COST', 'NO_MATCH_GCP_COST'
  cluster_id    STRING,
  job_id        STRING,
  run_id        STRING,
  usage_date    DATE,
  currency      STRING,
  error_detail  STRING,   -- free-text description
  raw_record    STRING,   -- JSON serialization of the problematic row
  created_at    TIMESTAMP
)
CLUSTER BY AUTO
```

```python
dbutils.notebook.exit(f"{catalog}.{schema}.dbspend360_error_log")
```
