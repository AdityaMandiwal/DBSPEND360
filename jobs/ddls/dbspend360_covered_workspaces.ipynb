# DDL: `dbspend360_covered_workspaces`

Creates the covered-workspace map populated by `dbspend360_covered_workspaces_app`.
Each row is a Databricks workspace whose data-plane managed resource group was
discovered in the queried Azure subscription via ARM.

**Widgets**
- `catalog` - target Unity Catalog name
- `schema`  - target schema name within `catalog`

```python
dbutils.widgets.text("catalog", "", "Catalog")
dbutils.widgets.text("schema", "", "Schema")
```

```python
catalog = dbutils.widgets.get("catalog").strip()
schema = dbutils.widgets.get("schema").strip()

if not catalog or not schema:
    raise ValueError("Both `catalog` and `schema` widgets must be set.")

spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")
```

```python
%sql
CREATE TABLE IF NOT EXISTS ${catalog}.${schema}.dbspend360_covered_workspaces (
  workspace_id    STRING,
  workspace_name  STRING,
  subscription_id STRING,
  workspace_url   STRING,
  updated_at      TIMESTAMP
)
CLUSTER BY AUTO
```

```python
dbutils.notebook.exit(f"{catalog}.{schema}.dbspend360_covered_workspaces")
```
